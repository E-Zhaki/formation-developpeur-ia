# Listes

Les listes en Python représentent des séquences ordonnées de valeurs. Voici un exemple de la façon de les créer :

```python
primes = [2, 3, 5, 7]
```

On peut mettre d'autres types de choses dans des listes :

```python
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
```

Nous pouvons même dresser une liste de listes :

```python
hands = [
    ['J', 'Q', 'K'],
    ['2', '2', '2'],
    ['6', 'A', 'K'], # (Comma after the last element is optional)
]
# (I could also have written this on one line, but it can get hard to read)
hands = [['J', 'Q', 'K'], ['2', '2', '2'], ['6', 'A', 'K']]
```

Une liste peut contenir un mélange de différents types de variables :

```python
my_favourite_things = [32, 'raindrops on roses', help]
# (Yes, Python's help function is *definitely* one of my favourite things)
```

## Indexation

Vous pouvez accéder à des éléments de liste individuels avec des crochets.

Quelle est la planète la plus proche du soleil ? Python utilise une indexation basée sur zéro, le premier élément a donc l'index 0.

```python
planets[0]
```

Quelle est la planète suivante, la plus proche du soleil ?

```python
planets[1]
```

Quelle est la planète la plus éloignée du soleil ?

Les éléments situés à la fin de la liste sont accessibles avec des nombres négatifs, à partir de -1 :

```python
planets[-1]
```

```python
planets[-2]
```

## Découpages

Quelles sont les trois premières planètes ? Nous pouvons répondre à cette question en utilisant le découpage :

```python
planets[0:3]
```

`planets[0:3]` est notre façon de demander les éléments de `planets`
 à partir de l'indice 0 et jusqu'à l'indice 3 non compris.

Les indices de début et de fin sont tous deux facultatifs. Si je n'indique pas l'indice de départ, il est supposé être 0. Je pourrais donc réécrire l'expression ci-dessus comme suit :

```python
planets[:3]
```

Si j'omets l'indice de fin, on suppose qu'il s'agit de la longueur de la liste.

```python
planets[3:]
```

C'est-à-dire que l'expression ci-dessus signifie "donnez-moi toutes les planètes à partir de l'indice 3".

Nous pouvons également utiliser des indices négatifs lors du découpage :

```python
# All the planets except the first and last
planets[1:-1]
```

```python
# The last 3 planets
planets[-3:]
```

## Modifications des listes

Les listes sont "mutables", ce qui signifie qu'elles peuvent être modifiées "sur place".

Une façon de modifier une liste est d'assigner une ou des valeurs à un index ou à un découpage.

Par exemple, disons que nous voulons renommer Mars :

```python
planets[3] = 'Malacandra'
planets
```

Hum, ça fait beaucoup de mots. Compensons en raccourcissant les noms des 3 premières planètes.

```python
planets[:3] = ['Mur', 'Vee', 'Ur']
print(planets)
# That was silly. Let's give them back their old names
planets[:4] = ['Mercury', 'Venus', 'Earth', 'Mars',]
```

## Fonctions utiles

Python possède plusieurs fonctions utiles pour travailler avec des listes.

`len` donne la longueur d'une liste :

```python
# How many planets are there?
len(planets)
```

`sorted` renvoie une version triée d'une liste :

```python
# The planets sorted in alphabetical order
sorted(planets)
```

`sum` fait ce à quoi on peut s'attendre :

```python
primes = [2, 3, 5, 7]
sum(primes)
```

Nous avons précédemment utilisé les fonctions `min` et `max` pour obtenir le minimum ou le maximum de plusieurs arguments. Mais nous pouvons également passer un seul argument de liste.

```python
max(primes)
```

## Interlude : la notion d'objets

Nous avons déjà utilisé le terme "objet" jusqu'à présent - vous avez peut-être même lu que tout en Python est un objet. Mais qu'est-ce que cela signifie ?

En bref, les objets transportent certaines choses avec eux. Vous accédez à ces choses en utilisant la syntaxe point de Python.

Par exemple, les nombres en Python transportent une variable associée appelée `imag` représentant leur partie imaginaire. (Vous n'aurez probablement jamais besoin de l'utiliser, sauf si vous faites des maths très bizarres).

```python
x = 12
# x is a real number, so its imaginary part is 0.
print(x.imag)
# Here's how to make a complex number, in case you've ever been curious:
c = 12 + 3j
print(c.imag)
```

Les choses qu'un objet transporte peuvent également inclure des fonctions. Une fonction attachée à un objet est appelée une **méthode**. (Les éléments non fonctionnels attachés à un objet, comme `imag`, sont appelés **attributs**).

Par exemple, les nombres ont une méthode appelée `bit_length`. Là encore, nous y accédons à l'aide de la syntaxe point :

```python
x.bit_length
```

Pour l'appeler réellement, nous ajoutons des parenthèses :

```python
x.bit_length()
```

De la même manière que nous pouvons passer des fonctions à la fonction `help` (par exemple `help(max)`), nous pouvons également passer des méthodes :

```python
help(x.bit_length)
```

Les exemples ci-dessus étaient un peu étranges. Aucun des types d'objets que nous avons étudiés jusqu'à présent (nombres, fonctions, booléens) ne possède d'attributs ou de méthodes que vous êtes susceptibles d'utiliser un jour.

Mais il s'avère que les listes ont plusieurs méthodes que vous utiliserez tout le temps.

## Méthodes utiles

`list.append` modifie une liste en ajoutant un élément à la fin :

```python
# Pluto is a planet darn it!
planets.append('Pluto')
```

Pourquoi la cellule ci-dessus n'a-t-elle pas de sortie ? Vérifions la documentation en appelant `help(planets.append)`.

```python
help(planets.append)
```

Ce message nous indique que `list.append` ne retourne rien. Mais si nous vérifions la valeur de `planets`, nous pouvons voir que l'appel de la méthode a modifié la valeur de `planets` :

```python
planets
```

`list.pop` supprime et renvoie le dernier élément d'une liste :

```python
planets.pop()
```

```python
planets
```

## Recherche dans les listes

Où se situe la Terre dans l'ordre des planètes ? Nous pouvons obtenir son indice en utilisant la méthode `list.index`.

```python
planets.index('Earth')
```

Elle arrive en troisième position (l'indexation commence à partir de zéro !).

A quel indice se trouve Pluton ?

```python
planets.index('Pluto')
```

Eh oui, c'est vrai, nous l'avions effacée quelques lignes plus haut...

Pour éviter les mauvaises surprises de ce genre, nous pouvons utiliser l'opérateur `in` pour déterminer si une liste contient une valeur particulière :

```python
# Is Earth a planet?
"Earth" in planets
```

```python
# Is Calbefraques a planet?
"Calbefraques" in planets
```

Il existe quelques autres méthodes de liste intéressantes que nous n'avons pas couvertes. Si vous voulez connaître toutes les méthodes et tous les attributs attachés à un objet particulier, vous pouvez appeler `help()` sur l'objet lui-même. Par exemple, `help(planets)` nous renseignera sur toutes les méthodes de la liste :

```python
help(planets)
```

Les listes comportent de nombreuses méthodes aux noms étranges comme `__eq__` et `__iadd__`. Ne vous en préoccupez pas trop pour l'instant. (Vous n'appellerez probablement jamais ces méthodes directement. Mais elles sont appelées en coulisse lorsque nous utilisons des syntaxes comme l'indexation ou les opérateurs de comparaison). Les méthodes les plus intéressantes se trouvent vers le bas de la liste (`append`, `clear`, `copy`, etc.).

## Tuples

Les tuples sont presque exactement les mêmes que les listes. Ils ne diffèrent que sur deux points.

**1** : La syntaxe pour les créer utilise des parenthèses au lieu de crochets.

```python
t = (1, 2, 3)
```

```python
t = 1, 2, 3 # equivalent to above
t
```

**2** : Ils ne peuvent pas être modifiés (ils sont immuables).

```python
t[0] = 100
```

Les tuples sont souvent utilisés pour les fonctions qui ont plusieurs valeurs de retour.

Par exemple, la méthode `as_integer_ratio()` des objets de type flottants renvoie un numérateur et un dénominateur sous la forme d'un tuple :

```python
x = 0.125
x.as_integer_ratio()
```

Ces valeurs de retour multiples peuvent être attribuées individuellement comme suit :

```python
numerator, denominator = x.as_integer_ratio()
print(numerator / denominator)
```
