# Introduction

Souvent, les données nous parviennent avec des noms de colonnes, des noms d'index ou d'autres conventions de dénomination dont nous ne sommes pas satisfaits. Dans ce cas, vous apprendrez à utiliser les fonctions de Pandas pour changer les noms des entrées incriminées en quelque chose de meilleur.

Vous découvrirez également comment combiner des données provenant de plusieurs DataFrames et/ou Series.

```python
import pandas as pd

reviews = pd.read_csv("../wine-reviews/winemag-data-130k-v2.csv", index_col=0)
```

# Renommer

La première fonction que nous allons introduire ici est `rename()`, qui vous permet de changer les noms d'index et/ou les noms de colonnes. Par exemple, pour changer la colonne `points` en `score`, nous faisons :

```python
reviews.rename(columns={'points': 'score'})
```

`rename()` vous permet de renommer les valeurs d'un index ou d'une colonne en spécifiant respectivement les mot-clés `index` ou `column`. Il supporte une variété de formats d'entrée, mais généralement un dictionnaire Python est le plus pratique. Voici un exemple d'utilisation pour renommer certains éléments de l'index.

```python
reviews.rename(index={0: 'firstEntry', 1: 'secondEntry'})
```

Vous renommerez probablement très souvent des colonnes, mais très rarement des valeurs d'index. Pour cela, `set_index()` est généralement plus pratique.

L'index de ligne et l'index de colonne peuvent tous deux avoir leur propre attribut `name`. La méthode complémentaire `rename_axis()` peut être utilisée pour changer ces noms. Par exemple :

```python
reviews.rename_axis("wines", axis='rows').rename_axis("fields", axis='columns')
```

# Combiner

Lorsque nous effectuons des opérations sur un ensemble de données, nous devons parfois combiner différents DataFrames et/ou Series de manière non triviale. Pandas dispose de trois méthodes principales pour ce faire. Par ordre de complexité croissante, ce sont `concat()`, `join()` et `merge()`. La plupart de ce que `merge()` peut faire peut également être fait plus simplement avec `join()`, nous l'omettrons donc et nous nous concentrerons sur les deux premières fonctions ici.

La méthode de combinaison la plus simple est `concat()`. Étant donné une liste d'éléments, cette fonction va écraser ces éléments ensemble le long d'un axe.

Ceci est utile lorsque nous avons des données dans différents objets DataFrame ou Series mais ayant les mêmes champs (colonnes). Un exemple : le jeu de données YouTube Videos, qui divise les données en fonction du pays d'origine (par exemple, le Canada et le Royaume-Uni, dans cet exemple). Si nous voulons étudier plusieurs pays simultanément, nous pouvons utiliser `concat()` pour les concaténer :

```python
canadian_youtube = pd.read_csv("../youtube-new/CAvideos.csv")
british_youtube = pd.read_csv("../youtube-new/GBvideos.csv")

pd.concat([canadian_youtube, british_youtube])
```

Le combinateur le plus complexe est `join()`. `join()` vous permet de combiner différents objets DataFrame qui ont un index en commun. Par exemple, pour extraire des vidéos dont la tendance est la même le même jour au Canada et au Royaume-Uni, nous pourrions faire ce qui suit :

```python
left = canadian_youtube.set_index(['title', 'trending_date'])
right = british_youtube.set_index(['title', 'trending_date'])

left.join(right, lsuffix='_CAN', rsuffix='_UK')
```

Les paramètres `lsuffix` et `rsuffix` sont nécessaires ici parce que les données ont les mêmes noms de colonne dans les ensembles de données britanniques et canadiens. Si ce n'était pas le cas (parce que, par exemple, nous les avions renommés au préalable), nous n'en aurions pas besoin.
