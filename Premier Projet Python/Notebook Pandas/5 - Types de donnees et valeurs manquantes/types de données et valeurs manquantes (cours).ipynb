# Introduction

Dans ce tutoriel, vous apprendrez à étudier les types de données dans une DataFrame ou une Series. Vous apprendrez également comment trouver et remplacer des entrées.

```python
import pandas as pd

reviews = pd.read_csv("../wine-reviews/winemag-data-130k-v2.csv", index_col=0)
```

# Dtypes

Le type de données pour une colonne dans un DataFrame ou une Series est connu sous le nom de `dtype`.

Vous pouvez utiliser la propriété `dtype` pour saisir le type d'une colonne spécifique. Par exemple, nous pouvons obtenir le dtype de la colonne `price` dans la DataFrame `reviews` :

```python
reviews.price.dtype
```

Sinon, la propriété `dtypes` renvoie le `dtype` de chaque colonne dans le DataFrame :

```python
reviews.dtypes
```

Les types de données nous renseignent sur la façon dont Pandas stockent les données en interne. `float64` signifie qu'il utilise un nombre à virgule flottante de 64 bits ; `int64` signifie un entier encodé sur 64 bits, etc.

Une particularité qu'il faut garder à l'esprit (et qui est très clairement affichée ici) est que les colonnes composées entièrement de chaînes de caractères n'ont pas leur propre type ; on leur donne plutôt le type `object`.

Il est possible de convertir une colonne d'un type en un autre là où une telle conversion a un sens en utilisant la fonction `astype()`. Par exemple, nous pouvons transformer la colonne `points` de son type de données `int64`existant en un type de données `float64` :

```python
reviews.points.astype('float64')
```

Un index DataFrame ou Series a également son propre `dtype` :

```python
reviews.index.dtype
```

Pandas prend également en charge des types de données plus exotiques, telles que les données catégorielles et les données chronologiques. Ces types de données sont plus rarement utilisés.

# Valeurs manquantes

Les valeurs manquantes des entrées reçoivent la valeur `NaN`, abréviation de "Not a Number". Pour des raisons techniques, ces valeurs `NaN` sont toujours du type `dtype` `float64`.

Pandas fournit quelques méthodes spécifiques aux données manquantes. Pour sélectionner les entrées `NaN`, vous pouvez utiliser `pd.isnull()` (ou son compagnon `pd.notnull()`). Cette méthode est destinée à être utilisée de cette manière :

```python
reviews[pd.isnull(reviews.country)]
```

Le remplacement des valeurs manquantes est une opération courante. Pandas propose une méthode très pratique pour ce problème : `fillna()`. `fillna()` propose quelques stratégies différentes pour atténuer ce type de données. Par exemple, nous pouvons simplement remplacer chaque `NaN` par `"Unknown"` (inconnu) :

```python
reviews.region_2.fillna("Unknown")
```

Il se peut aussi que nous ayons une valeur non nulle que nous voudrions remplacer. Par exemple, supposons que depuis que cet ensemble de données a été publié, la critique Kerin O'Keefe a changé son identifiant Twitter de `@kerinokeefe` à `@kerino`. Une façon de refléter cela dans l'ensemble de données est d'utiliser la méthode `replace()` :

```python
reviews.taster_twitter_handle.replace("@kerinokeefe", "@kerino")
```

La méthode `replace()` mérite d'être mentionnée ici car elle est pratique pour remplacer des données manquantes auxquelles on donne une sorte de valeur sentinelle dans l'ensemble de données : des choses comme "Unknown" (inconnu), "Undisclosed" (non divulgué), "Invalid" (invalide), et ainsi de suite.
