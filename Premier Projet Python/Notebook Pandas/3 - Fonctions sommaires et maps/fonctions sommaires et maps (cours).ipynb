# Fonctions sommaires

Pandas offre de nombreuses "fonctions sommaires" simples qui restructurent les données de manière utile. Prenons par exemple la méthode `describe()` :

```python
reviews.points.describe()
```

Cette méthode génère un résumé de haut niveau des attributs de la colonne donnée. Elle tient compte du type, ce qui signifie que sa sortie change en fonction du type de données de l'entrée. La sortie ci-dessus n'a de sens que pour les données numériques ; pour les données de type chaîne, voici ce que nous obtenons :

```python
reviews.taster_name.describe()
```

Si vous souhaitez obtenir une statistique sommaire simple et particulière sur une colonne d'un DataFrame ou d'une Series, il existe généralement une fonction pandas utile qui permet de le faire.

Par exemple, pour connaître la moyenne des points attribués (par exemple, la note moyenne d'un vin), on peut utiliser la fonction `mean()` :

```python
reviews.points.mean()
```

Pour voir une liste de valeurs uniques, nous pouvons utiliser la fonction `unique()` :

```python
reviews.taster_name.unique()
```

Pour voir une liste de valeurs uniques et leur fréquence dans l'ensemble de données, nous pouvons utiliser la méthode `value_counts()` :

```python
reviews.taster_name.value_counts()
```

# Maps

Une "map" est un terme, emprunté aux mathématiques, pour désigner une fonction qui prend un ensemble de valeurs et les "fait correspondre" à un autre ensemble de valeurs. En science des données, nous avons souvent besoin de créer de nouvelles représentations à partir de données existantes, ou de transformer les données du format dans lequel elles se trouvent actuellement au format dans lequel nous voulons qu'elles se trouvent plus tard. Ce sont les "map" qui gèrent ce travail, ce qui les rend extrêmement importantes.

Il existe deux méthodes de "mapping" que vous utiliserez souvent.

`map()` est la première, et légèrement plus simple. Supposons par exemple que nous voulions ramener à 0 la moyenne des notes des vins reçus, ce que nous pouvons faire comme suit :

```python
review_points_mean = reviews.points.mean()
reviews.points.map(lambda p: p - review_points_mean)
```

La fonction que vous passez à `map()` doit attendre une seule valeur de la Series (une valeur ponctuelle, dans l'exemple ci-dessus), et retourner une version transformée de cette valeur. `map()` retourne une nouvelle Series où toutes les valeurs ont été transformées par votre fonction.

`apply()` est la méthode équivalente si nous voulons transformer un DataFrame entier en appelant une méthode personnalisée sur chaque ligne.

```python
def remean_points(row):
    row.points = row.points - review_points_mean
    return row

reviews.apply(remean_points, axis='columns')
```

Si nous avions appelé `reviews.apply()` avec `axis='index'`, alors au lieu de passer une fonction pour transformer chaque ligne, nous aurions dû donner une fonction pour transformer chaque colonne.

Notez que `map()` et `apply()` renvoient respectivement de nouvelles Series et DataFrames transformées. Ils ne modifient pas les données originales auxquelles ils font appel. Si l'on regarde la première ligne de `reviews`, on peut voir qu'elle a toujours sa valeur `points` d'origine.

```python
reviews.head(1)
```

Pandas fournit de nombreuses opérations de "mapping" courantes. Par exemple, voici un moyen plus rapide de modifier notre colonne de points :

```python
review_points_mean = reviews.points.mean()
reviews.points - review_points_mean
```

Dans ce code, nous effectuons une opération entre un grand nombre de valeurs sur la gauche (tout ce qui se trouve dans la série) et une seule valeur sur la droite (la valeur moyenne). Pandas examine cette expression et découvre que nous devons avoir l'intention de soustraire cette valeur moyenne de chaque valeur de l'ensemble de données.

Pandas comprendra également ce qu'il faut faire si nous effectuons ces opérations entre des Series de longueur égale. Par exemple, une façon simple de combiner les informations sur les pays et les régions dans l'ensemble de données serait de procéder comme suit :

```python
reviews.country + " - " + reviews.region_1
```

Ces opérateurs sont plus rapides que `map()` ou `apply()` car ils utilisent des accélérateurs intégrés dans Pandas. Tous les opérateurs Python standard (`>`, `<`, `==`, etc.) fonctionnent de cette manière.

Cependant, ils ne sont pas aussi flexibles que `map()` ou `apply()`, qui peuvent faire des choses plus avancées, comme appliquer une logique conditionnelle, ce qui ne peut pas être fait avec l'addition et la soustraction seules.
