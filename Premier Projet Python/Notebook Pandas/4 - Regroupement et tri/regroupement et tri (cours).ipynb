# Introduction

Les Maps nous permettent de transformer les données dans un DataFrame ou une Series, une valeur à la fois pour une colonne entière. Cependant, nous voulons souvent regrouper nos données, puis faire quelque chose de spécifique au groupe dans lequel les données se trouvent.

Cela se fait avec l'opération `groupby()`. Nous aborderons également d'autres sujets, tels que des moyens plus complexes d'indexer vos DataFrames, ainsi que la manière de trier vos données.

```python
import pandas as pd

reviews = pd.read_csv("../wine-reviews/winemag-data-130k-v2.csv", index_col=0)
```

# Analyse par groupe

Une fonction que nous avons beaucoup utilisée jusqu'à présent est la fonction `value_counts()`. Nous pouvons reproduire ce que fait la fonction `value_counts()` en procédant comme suit :

```python
reviews.groupby('points').points.count()
```

`groupby()` a créé un groupe de `reviews` qui attribuaient les mêmes valeurs de points aux vins donnés. Ensuite, pour chacun de ces groupes, nous avons saisi la colonne `points()` et compté combien de fois elle apparaissait. `value_counts()` n'est qu'un raccourci vers cette opération `groupby()`.

Nous pouvons utiliser n'importe laquelle des fonctions sommaires que nous avons utilisées auparavant avec ces données. Par exemple, pour obtenir le vin le moins cher dans chaque catégorie de valeur en points, nous pouvons faire ce qui suit :

```python
reviews.groupby('points').price.min()
```

Vous pouvez voir chaque groupe que nous générons comme étant une tranche de notre DataFrame contenant uniquement des données dont les valeurs correspondent. Ce DataFrame nous est accessible directement en utilisant la méthode `apply()`, et nous pouvons ensuite manipuler les données comme bon nous semble. Par exemple, voici une façon de sélectionner le nom du premier vin examiné dans chaque vignoble de l'ensemble de données :

```python
reviews.groupby('winery').apply(lambda df: df.title.iloc[0])
```

Pour un contrôle encore plus fin, vous pouvez également regrouper par plusieurs colonnes. Voici, à titre d'exemple, comment nous pourrions sélectionner le meilleur vin par pays et par province :

```python
reviews.groupby(['country', 'province']).apply(lambda df: df.loc[df.points.idxmax()])
```

Une autre méthode `groupby()` qui mérite d'être mentionnée est `agg()`, qui vous permet d'exécuter un tas de fonctions différentes sur votre DataFrame simultanément. Par exemple, nous pouvons générer un simple résumé statistique de l'ensemble de données comme suit :

```python
reviews.groupby(['country']).price.agg([len, min, max])
```

L'utilisation efficace de `groupby()` vous permettra de faire beaucoup de choses vraiment puissantes avec votre ensemble de données.

# Multi-indexes

Dans tous les exemples que nous avons vus jusqu'à présent, nous avons travaillé avec des objets DataFrame ou Series ayant un indice/index unique. `groupby()` est légèrement différent dans le fait que, selon l'opération que nous effectuons, il en résultera parfois ce que l'on appelle un multi-index.

Un multi-index diffère d'un index normal en ce qu'il comporte plusieurs niveaux. Par exemple :

```python
countries_reviewed = reviews.groupby(['country', 'province']).description.agg([len])
countries_reviewed
```

```python
mi = countries_reviewed.index
type(mi)
```

Les multi-indexes ont plusieurs méthodes pour traiter de leur structure à niveaux qui sont absentes pour les indices à niveau unique. Ils nécessitent également deux niveaux d'étiquettes pour récupérer une valeur.

Les cas d'utilisation d'un multi-index sont détaillés ainsi que les instructions d'utilisation dans la section `MultiIndex / Advanced Selection` de la documentation Pandas.

Toutefois, en général, la méthode multi-index que vous utiliserez le plus souvent est celle qui permet de reconvertir un index régulier, la méthode `reset_index()` :

```python
countries_reviewed.reset_index()
```

# Tri

En examinant à nouveau les `countries_reviewed`, nous pouvons constater que le regroupement renvoie les données par ordre d'indice, et non par ordre de valeur. C'est-à-dire que lors de la production du résultat d'un `groupby`, l'ordre des lignes dépend des valeurs de l'indice, et non des données.

Pour obtenir des données dans l'ordre souhaité, nous pouvons les trier nous-mêmes. La méthode `sort_values()` est pratique pour cela.

```python
countries_reviewed = countries_reviewed.reset_index()
countries_reviewed.sort_values(by='len')
```

`sort_values()` propose par défaut un tri ascendant, où les valeurs les plus faibles passent en premier. Cependant, la plupart du temps, nous voulons un tri descendant, où les valeurs les plus élevées passent en premier. C'est ainsi que cela se passe :

```python
countries_reviewed.sort_values(by='len', ascending=False)
```

Pour trier par valeurs d'indice, utilisez la méthode compagnon `sort_index()`. Cette méthode a les mêmes arguments et le même ordre par défaut :

```python
countries_reviewed.sort_index()
```

Enfin, sachez que vous pouvez trier par plus d'une colonne à la fois :

```python
countries_reviewed.sort_values(by=['country', 'len'])
```
